```python
import tensorflow as tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
X = iris.data

y = iris.target.reshape(-1,1)
print(X)
print(y)
```

```python
encoder = OneHotEncoder(sparse_output=False)
y_encoded = encoder.fit_transform(y)

print(y_encoded)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
X_train_nn, X_test_nn, y_train_nn, y_test_nn = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
```

```python
#Machine learning
tree_model = DecisionTreeClassifier()
tree_model.fit(X_train, y_train)

y_pred_tree = tree_model.predict(X_test)
tree_accuracy = accuracy_score(y_test, y_pred_tree)
print("Decision Tre Accuracy (ML):", tree_accuracy)
```

```python
#deep learning
#neural network

nn_model = Sequential([
    Dense(16, activation='relu', input_shape=(4,)),
    Dense(16, activation='relu'),
    Dense(3, activation='softmax')
])


nn_model.compile(optimizer='adam',
loss='categorical_crossentropy',
metrics=['accuracy'])

nn_model.fit(X_train_nn, y_train_nn, epochs=50, batch_size=5, validation_split=0.1, verbose=1)
#epochs = number of times the model sees the entire dataset
#batch_size = the number of samples which are used at once to X_train_nn
#validation_split = the fraction of the dataset that will be used for validation_split
#verbose=1 = this will print progress bars and training information during training

nn_loss, nn_accuracy = nn_model.evaluate(X_test_nn, y_test_nn, verbose=0)
print("Neural Network Accuracy (DL):", nn_accuracy)
```

```python
#comparison of results

print("\nComparison of ML and DL Approaches:")
print("Decision Tree Accuracy (ML):", tree_accuracy)
print("Neural Network Accuracy (DL):", nn_accuracy)
```
